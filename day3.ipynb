```python
import pandas as pd
df = pd.read_csv("sales.csv")
print(df)
print(df.info())
print(df.isnull().sum())
df=df.drop_duplicates()
df['Profit'] = df['Profit'].fillna(df['Profit'].mean())
print(df)
```

```python
import pandas as pd
df = pd.read_csv("salary.csv")
print(df)
avg_salary = df.groupby('Department')['Salary'].mean()
print(avg_salary)
summary = df.groupby('Department')['Salary'].agg(    ['mean', 'max', 'min', 'count']
)
print(summary)
```

```python
numbers = [10,20,30,40,50]
mean = sum(numbers)/len(numbers)
print("Mean=",mean)
```

```python
numbers = [10,20,30,40,50]
numbers.sort()
n = len(numbers)
if n % 2==0:
    median = (numbers[n//2-1]+numbers[n//2]/2)
else:
    median = numbers[n//2]
print("Median =",median)
```

```python
numbers=[10,20,20,30,40,20,50]
mode = max(set(numbers),key=numbers.count)
print("Mode",mode)
```

```python
numbers = [10.20,30,40,50]
mean = sum(numbers)/len(numbers)
variance = sum((x-mean)**2 for x in numbers)/len(numbers)
print("Variance=",variance)
```

```python
numbers=[10,20,20,30,40,20,50]
mean = sum(numbers)/len(numbers)
variance=sum((x-mean)**2 for x in numbers)/len(numbers)
std_deviation= variance**0.5
print("std deviation is ",std_deviation)
```

```python
import statistics as st
data=[10,20,30,40,50]
print("Mean:",st.mean(data))
print("Median:",st.median(data))
print("Mode:",st.mode(data))
print("Variance:",st.variance(data))
print("Standard Deviation::",st.stdev(data))
```

```python
import math
num = 25
print("Squre root:",math.sqrt(num))
print("power (5^2):", math.pow(5,2))
print("Factorial of 5:", math.factorial(5))
print("Value of Pi:", math.pi)
print("Value of e:", math.e)
print("Ceil of 4.3:", math.ceil(4.3))
print("Floor of 4.7:", math.floor(4.7))
```

```python
from scipy import stats
data = [10, 20, 30, 40, 50]
print("Mean:",stats.tmean(data))
print("Variance:",stats.tvar(data))
print("Standard Deviation:", stats.tstd(data))
```

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Array:", arr)
print("Sum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Maximum:", np.max(arr))
print("Minimum:", np.min(arr))
```
